```python
df["Travellers"] = df["Adults"] + df["Children"]
```

```python
df["Grossvalue"] = df["UnitPriceUSD"]
```

```python
df["DiscountAmount"] = df["Grossvalue"] - df["DiscountPct"]
```

```python
#filter/classify travellers by their booking sizes
df["BookingCategory"] = np.where(
df["Travellers"] >= 4,
"Kubwa", 
"Ndogo")
```

```python
conditions = [
    df["DiscountPct"] == 0,
    df["DiscountPct"] <= 0.10,
    df["DiscountPct"] > 0.10,
]
choices = ["No discount", "Low", "High"]

df["DiscountTier"] = np.select(conditions, choices, default="Unknown")
df.head()
```

```python
#Answering Busines questions?
# 1. Is the Business growing?

BusinessGrowth = (
    df.groupby("BookingYear").agg(
        Total_net_value = ("NetValue","sum"),
        Travellers_Served = ("Travellers","sum"),
        Booking_made = ("BookingID","nunique")
    )
    .reset_index()
)
BusinessGrowth.head()
```
